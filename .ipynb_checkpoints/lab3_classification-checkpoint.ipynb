## Q1:(3 pts) Creating Label Column

#### 1.1(2 pts) Create a new categorical column named "DelayGroup" based on the "DepDelayMinutes" column in the dataframe. The "DelayGroup" should be assigned according to the following conditions:


1. If the departure delay is exactly 0 minutes, label it as "OnTime_Early".
2. If the departure delay is greater than 0 minutes and less than or equal to 15 minutes, label it as "Small_Delay".
3. If the departure delay is greater than 15 minutes and less than or equal to 45 minutes, label it as "Medium_Delay".
4. If the departure delay is greater than 45 minutes, label it as "Large_Delay".
5. If the flight was cancelled, label it as 'Cancelled'.
6. If none of the above conditions are met, label it as 'Unknown'.

Don't Forget to Drop the DepDelayMinutes


```python
conditions = [ flightdf['Cancelled'] == True, 
              flightdf['DepDelayMinutes'] == 0, 
              (flightdf['DepDelayMinutes'] > 0) & (flightdf['DepDelayMinutes'] <= 15), 
              (flightdf['DepDelayMinutes'] > 15) & (flightdf['DepDelayMinutes'] <= 45), 
              (flightdf['DepDelayMinutes'] > 45) ]

labels = ['Cancelled', 'OnTime_Early', 'Small_Delay', 'Medium_Delay', 'Large_Delay']

flightdf['DelayGroup'] = np.select(conditions, labels, default='Unknown')
flightdf['DelayGroup'] = flightdf['DelayGroup'].astype('category')
flightdf.drop(columns=['DepDelayMinutes'], inplace=True)
```

#### 1.2(1 pt)Print the counts of different categories in the "DelayGroup" column

```python
flightdf['DelayGroup'].value_counts()
```

There is no unknown delay group, so all records are categorized correctly based on the conditions provided. But, there is a huge class imbalance in our label 'DelayGroup' .

## Q2.(7 pts) Feature Encoding


#### 2.1 (3 pts) Encode your feature categorical columns using label encoder

```python
label_encoders = {}

for column in flightdf.select_dtypes(include=['object']).columns:
    label_encoders[column] = LabelEncoder()
    flightdf[column] = label_encoders[column].fit_transform(flightdf[column])

print("Data types after encoding:")
print(flightdf.dtypes)
```

#### 2.2 (2 pts) Handle missing values by using fillna

```python
print("Missing values per column:")
print(flightdf.isnull().sum())
print("-" * 50)

not_cat_cols = flightdf.select_dtypes(exclude=['category']).columns
flightdf[not_cat_cols] = flightdf[not_cat_cols].fillna(0)
print("Missing values after fillna:")
print(flightdf.isnull().sum().sum()) # Total NaN values after filling
```

#### 2.3 (2 pts) Normalize the features using Standard Scaler

```python
X = flightdf.drop(columns=['DelayGroup'])
y = flightdf['DelayGroup']

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print(f"Shape of scaled features: {X_scaled.shape}")
print(f"Shape of target variable: {y.shape}")
```

## Q3.(15 pts, 3 pts each) PCA and K-Means

#### 3.1 Perform PCA on the feature columns using 2 components

```python
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

print(f"Shape before PCA: {X_scaled.shape}")
print(f"Shape after PCA: {X_pca.shape}")
print(f"Variance explained by PCA components: {pca.explained_variance_ratio_.sum():.2%}")
```

Variance is not super high but expected since we had 30 features and now we have 2 features. We lost %65.39 information. However, we can visualize it now.

#### 3.2 Perform Elbow method for finding best K on the PCA features

```python
k_values = range(1, 11)
inertia = []

for k in k_values:
    kmeans = KMeans(n_clusters=k, n_init=10, random_state=42)
    kmeans.fit(X_pca)
    inertia.append(kmeans.inertia_)

plt.figure(figsize=(10, 5))
plt.plot(k_values, inertia, 'bx-')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('Inertia')
plt.title('Elbow Method for Optimal K')
plt.xticks(k_values)
plt.show()
```

Even though, we have 5 different delay groups. The best number of cluster is 4. I assume that, 'Cancelled' delay group was really small of the column that's why it merges them into other groups.

#### 3.3 Perform K-means Clustering on the PCA features for best K

```python
kmeans = KMeans(n_clusters=4, n_init=10, random_state=42)
kmeans.fit(X_pca)
cluster_labels = kmeans.labels_

print("Cluster counts:")
print(pd.Series(cluster_labels).value_counts())
```

#### 3.4 Create a visualization for clustering results using principal components 1 and 2.

```python
df_pca = pd.DataFrame(X_pca, columns=['PC1', 'PC2'])
df_pca['Cluster'] = cluster_labels

sns.scatterplot(x='PC1', y='PC2', hue='Cluster', data=df_pca, palette='viridis')
plt.title('KMeans Clustering using PCA')
plt.show()
```

#### 3.5 Create a visualization for delay groups using principal components 1 and 2.

```python
df_pca['DelayGroup'] = y.values  

sns.scatterplot(x='PC1', y='PC2', hue='DelayGroup', data=df_pca, palette='viridis')
plt.title('Delay Groups using PCA')
plt.legend(loc='upper right')
plt.show()
```

KMeans cluster doesn't match delay groups. KMeans is an unsupervised learning, it finds natural patterns in the data without knowing our delay labels.The clusters it found are based on different patterns than what defines our delay groups.

Additionally, PCA only explains 34.71% of the variance with 2 components. Meaning 65.29% of information is lost which makes it harder to seperate delay groups in 2D space.

## Q4.(15 pts, 3 pts each) Classification on before and after PCA dataset

### Classification before PCA

#### 4.1 Based on the feature column and label column split the data into training and test data

```python
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)

print(f"Training set shape: {X_train.shape}, {y_train.shape}")
print(f"Testing set shape: {X_test.shape}, {y_test.shape}")
```

#### 4.2 Apply Logisitic Regression and print accuracy and classification report

```python
lr = LogisticRegression(max_iter=1000, random_state=42)
lr.fit(X_train, y_train)

y_pred_lr = lr.predict(X_test)

print(f"Logistic Regression Accuracy: {accuracy_score(y_test, y_pred_lr):.4f}") # accuracy
print("Classification Report:")
print(classification_report(y_test, y_pred_lr)) # precision, recall, f1-score, support
```

### Classification with PCA

#### 4.3 Perform PCA on the previously scaled data and select components in the range 4-20

```python
pca_scores = {}

for n in range(4,21):
    # Fit PCA and transform the data
    pca =PCA(n_components=n)
    X_pca_train = pca.fit_transform(X_train)
    X_pca_test = pca.transform(X_test)

    # Train Logistic Regression on PCA-transformed data
    lr_pca = LogisticRegression(max_iter=1000, random_state=42)
    lr_pca.fit(X_pca_train, y_train)
    
    # Predict and evaluate
    y_pred_pca = lr_pca.predict(X_pca_test)
    pca_scores[n] = accuracy_score(y_test, y_pred_pca)
    print(f"PCA with {n} components - Accuracy: {pca_scores[n]:.4f}")

best_n = max(pca_scores, key=pca_scores.get)
print(f"Best number of PCA components: {best_n} with Accuracy: {pca_scores[best_n]:.4f}")

# Plot
plt.figure(figsize=(10, 5))
plt.plot(list(pca_scores.keys()), list(pca_scores.values()), 'bx-')
plt.xlabel('Number of PCA Components')
plt.ylabel('Accuracy')
plt.title('PCA Components vs Accuracy')
plt.xticks(range(4, 21))
plt.show()
```

#### 4.4 Apply Logisitic Regression and print accuracy and classification report

```python
pca_best = PCA(n_components=20)
X_train_pca = pca_best.fit_transform(X_train)
X_test_pca  = pca_best.transform(X_test)

lr_best = LogisticRegression(max_iter=1000, random_state=42)
lr_best.fit(X_train_pca, y_train)
y_pred_pca = lr_best.predict(X_test_pca)

print(f"Variance explained: {pca_best.explained_variance_ratio_.sum():.2%}")
print(classification_report(y_test, y_pred_pca))
```

#### 4.5 Give observation about which model performed better along with the reasoning

```python
print(f"""
Before PCA:
  Accuracy:          {accuracy_score(y_test, y_pred_lr):.4f}
  Variance Explained: 100% (no PCA applied, all 30 features used)

After PCA (n=20):
  Accuracy:           {accuracy_score(y_test, y_pred_pca):.4f}
  Variance Explained: {pca_best.explained_variance_ratio_.sum():.2%}
{'*-' * 30}

Observation:
The model before PCA performed slightly better (0.6478 vs 0.6400).
PCA retained 99.25% of variance but still lost some useful information.
Both models struggled with Medium_Delay and Small_Delay due to class imbalance.
The main problem is class imbalance, not PCA.
""")
```

## Q5.(16 pts, 2 pts each) Class Imbalance

```python
flightdf = pd.read_csv('data/flight200k.csv')
print(flightdf.shape)
print(flightdf['DelayGroup'].value_counts())
```

```python
print(f"Data types:\n{flightdf.dtypes}")
```

```python
flightdf.head()
```

#### 5.1 Split data into feature columns and label column where label column is DelayGroup

```python
X = flightdf.drop(columns=['DelayGroup'])
y = flightdf['DelayGroup']

X = X.fillna(0)

# Scale features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print("X shape:", X_scaled.shape)
print("\nClass distribution:")
print(y.value_counts())
```

```python
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```

#### 5.2 Perform Random Forest on the dataset

```python
rf = RandomForestClassifier(n_estimators=100, random_state=42)
rf.fit(X_train, y_train)
y_pred_rf = rf.predict(X_test)

print("Random Forest (baseline):")
print(classification_report(y_test, y_pred_rf))

acc_baseline = accuracy_score(y_test, y_pred_rf)
print(f"Baseline Accuracy: {acc_baseline:.4f}")
```

#### 5.3 Perform Over Sampling on the dataset and print the count of each class

Use RandomOverSampler present in the imblearn library

```python
ros = RandomOverSampler(random_state=42)
X_train_ros, y_train_ros = ros.fit_resample(X_train, y_train)

print("Class distribution after RandomOverSampler:")
print(pd.Series(y_train_ros).value_counts())
```

#### 5.4 Perform Random Forest on the oversampled dataset

```python
rf_ros = RandomForestClassifier(n_estimators=100, random_state=42) # RandomOverSampler on Random Forest
rf_ros.fit(X_train_ros, y_train_ros)
y_pred_ros = rf_ros.predict(X_test)

print("Random Forest (RandomOverSampler):")
print(classification_report(y_test, y_pred_ros))
acc_ros = accuracy_score(y_test, y_pred_ros)
print(f"Accuracy: {acc_ros:.4f}")
```

#### 5.5 Apply Under Sampling on the dataset and print the count of each class

Use RandomUnderSampler present in the imblearn library

```python
# Import RandomUnderSampler
from imblearn.under_sampling import RandomUnderSampler

# Apply RandomUnderSampler
rus = RandomUnderSampler(random_state=42)
X_train_rus, y_train_rus = rus.fit_resample(X_train, y_train)
print("Class distribution after RandomUnderSampler:")
print(pd.Series(y_train_rus).value_counts()) # Cancelled has the smallest class in the training set and it is 4768 (80% of 'Cancelled' class in the original training set)
```

#### 5.6 Perform Random Forest on the undersampled dataset

```python
rf_rus = RandomForestClassifier(n_estimators=100, random_state=42) # RandomUnderSampler on Random Forest
rf_rus.fit(X_train_rus, y_train_rus)
y_pred_rus = rf_rus.predict(X_test)

print("Random Forest (RandomUnderSampler):")
print(classification_report(y_test, y_pred_rus))
acc_rus = accuracy_score(y_test, y_pred_rus)
print(f"Accuracy: {acc_rus:.4f}")
```

#### 5.7 Apply SMOTE on the dataset to create new samples

```python
# Import SMOTE
from imblearn.over_sampling import SMOTE

smote = SMOTE(random_state=42)
X_train_smote, y_train_smote = smote.fit_resample(X_train, y_train)

print("Class distribution after SMOTE:")
print(pd.Series(y_train_smote).value_counts())
```

#### 5.8 Perform classification using Random Forest Classifier

```python
rf_smote = RandomForestClassifier(n_estimators=100, random_state=42)
rf_smote.fit(X_train_smote, y_train_smote)
y_pred_smote = rf_smote.predict(X_test)

print("Random Forest (SMOTE):")
print(classification_report(y_test, y_pred_smote))
acc_smote = accuracy_score(y_test, y_pred_smote)
print(f"Accuracy: {acc_smote:.4f}")
```

# 6. (4 pts) Report

####  Give observation about which model performed better along with the reasoning

| Method | Accuracy |
|--------|----------|
| Baseline RF | 0.7531 |
| OverSampler RF | 0.7645 |
| UnderSampler RF | 0.6199 |
| SMOTE RF | 0.7633 |

### Observations:

**1.Baseline Random Forest (0.7531)**
Strong baseline but struggled with minority classes. Small Delay recall = 0.20, Medium_Delay recall = 0.36.
Model heavily biased towards OnTime_Early class.

**2. RandomOverSampler (0.7645)**
Duplicating minority rows improved all minority classes.
Medium_Delay recall = 0.36 -> 0.50
Small_Delay recall = 0.20 -> 0.30

**3. RandomUnderSampler (0.6199)**
Lost %85 of training data and accuracy dropped from 0.75 to 0.62.
OnTime_Early recall dropped from 0.96 to 0.65.
Not recommended it for large datasets.

**4. SMOTE (0.7633)**
Very close to RandomOverSampler (0.7645).
On large datasets, synthetic sample add less value.

### Conclusion:
RandomOverSampler performed the best on this dataset.
RandomUnderSampler isn't recommended when losing majority class data hurts performance significantly.
SMOTE has less value on large datasets.
